```python
##Listaqde calificaciones del grupo a de probabilidades y estadisticas
from statistics import *
grupoa=[70,70,65,90,100,95,45,80,73,20,100,96]
print(grupoa)
type(grupoa)
```

Determinar la moda del ejemplo 1.
1.La moda es el valor que mas se repite dentro un conjunto de datos. 2.A diferencia de la media y mediana , la moda no requiere de valores numericos y puede utilizarse con datos de categorias o discretos.

```python
# Primero definimos la lista que faltaba para que no dé error
tiempos100mts = [15.1, 17.2, 15.1, 17.2, 18.0]

# Usamos la función median para determinar el tiempo promedio de la lista de valores
# Definamos una llamada mediana para calcular la mediana con la funcion median
media = median(tiempos100mts)
print("La mediana del tiempo es de=>", median(tiempos100mts))
```

```python
#Definamos una variable llamada moda poora calcular la moda de la lista con la funcion mode
moda=mode(tiempos100mts)
print(moda)
```

```python
#Definamos la funcion de multiples modas con multimode
multimoda=multimode(tiempos100mts)
print(multimoda)
```

Ejemplo 3 histogramas de 2 grupos.

```python
#Importamos libreria para generrar numeros aleatoria
import random
```

```python
fabrica1=[]
for i in range(100):
  numero_aleatorio=round(random.uniform(50,100),2)
  fabrica1.append(numero_aleatorio)
print(fabrica1)
```

```python
fabrica2=[]
for i in range(100):
  numero_aleatorio=round(random.uniform(70,100),2)
  fabrica2.append(numero_aleatorio)
print(fabrica2)
```

```python
#Importamos una libreria nueva pars grsficar valores
import matplotlib.pyplot as plt
```

```python
media1=mean(fabrica1)
mediana1=median(fabrica1)
moda1=mode(fabrica1)
print("Valores de la fabrica 1")
print("La media de la fabrica 1 es=>",media1)
print("La mediana de la fabrica 1 es=>",mediana1)
print("La moda de la fabrica 1 es=>",moda1)
```

```python
media2=mean(fabrica2)
mediana2=median(fabrica2)
moda2=mode(fabrica2)
print("Valores de la fabrica 2")
print("La media de la fabrica 2 es=>",media2)
print("La mediana de la fabrica 2 es=>",mediana2)
print("La moda de la fabrica 2 es=>",moda2)
```

```python
plt.hist(fabrica1)
plt.title("Produccion diaria d la Fabrica 1")
```

```python
plt.hist(fabrica2)
plt.title("Produccion diaria d la Fabrica 2")
```
